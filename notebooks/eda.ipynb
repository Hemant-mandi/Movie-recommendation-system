**Objective:** Analyze user viewing history, ratings, and genre interactions on an OTT platform to prepare for personalized movie/show recommendations.

```python
import sys
import os
sys.path.append(os.path.abspath('..'))

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from src.utils import load_movies, load_ratings, load_tags, clean_data, get_movie_stats, get_user_stats, save_processed_data

# Set visual style
sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams["figure.figsize"] = (10, 5)
```

## 1. Data Ingestion
Load the raw MovieLens datasets: `movies.csv`, `ratings.csv`, and `tags.csv`.

```python
movies_raw = load_movies("small")
ratings_raw = load_ratings("small")
tags_raw = load_tags("small")

print(f"Movies shape: {movies_raw.shape}")
print(f"Ratings shape: {ratings_raw.shape}")
print(f"Tags shape: {tags_raw.shape}")
```

## 2. Data Preprocessing & Feature Engineering
- Extract release year from title
- Parse multi-label pipe-separated genres
- Merge ratings with movie metadata

```python
movies, ratings, merged = clean_data(movies_raw, ratings_raw)
merged.head()
```

## 3. Exploratory Data Analysis (EDA)
### A. Rating Distribution Analysis
How are users rating content on the platform?

```python
plt.figure(figsize=(8, 4))
sns.countplot(data=ratings, x="rating", palette="mako")
plt.title("Distribution of User Ratings (0.5 to 5.0 Stars)", fontsize=14)
plt.xlabel("Rating Value")
plt.ylabel("Frequency Count")
plt.show()
```

### B. Genre Popularity Distribution

```python
genre_counts = movies["genres"].str.split("|").explode().value_counts()

plt.figure(figsize=(10, 5))
genre_counts.plot(kind="bar", color="crimson")
plt.title("Most Common Content Genres on Platform", fontsize=14)
plt.xlabel("Genre")
plt.ylabel("Number of Titles")
plt.xticks(rotation=45)
plt.show()
```

### C. Most Watched & Rated Titles

```python
movie_stats = get_movie_stats(merged)
print("Top 10 Most Rated Movies:")
display(movie_stats.head(10))
```

## 4. Export Processed Data

```python
save_processed_data(movies, ratings, merged)
print("Data cleaning and preparation complete!")
```
